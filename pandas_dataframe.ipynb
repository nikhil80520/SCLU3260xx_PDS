## 1. What is a DataFrame?

A DataFrame is a **2-dimensional labeled data structure** - like a spreadsheet or SQL table.

| Concept | What it is |
|---|---|
| **Full table** | DataFrame |
| **One column** | Series |
| **One row** | Also a Series (index = column names) |

```
         name    department   salary
0   Pooja D.    HR           55000
1   Tara M.     IT           72000
2   Kabir V.    Finance      61000
```

> **Key Insight:** Every column of a DataFrame is a Series. Most Series methods also work on DataFrame - you just need to specify **which column** or **which direction** (`axis`).


````python
# One column -> Series
df = pd.read_csv('employees.csv')

print(type(df['name']))        # <class 'pandas.core.series.Series'>
print(type(df.iloc[0]))        # <class 'pandas.core.series.Series'>
````

## 2. Creating a DataFrame

### 3 Ways to Create a DataFrame

| Method | Series | DataFrame |
|---|---|---|
| From List | `pd.Series([1,2,3])` | `pd.DataFrame([[..],[..]], columns=[...])` (2D list) |
| From Dictionary | keys = **index** | keys = **column names** |
| From CSV | `read_csv(...).squeeze("columns")` | `read_csv(...)` directly |

> **Common Confusion:** In a dictionary, keys become **index** for Series but **column names** for DataFrame.


````python
# Method 1: From a 2D list
student_data = [[100, 80, 10], [90, 70, 7], [120, 100, 14]]
df1 = pd.DataFrame(student_data, columns=['IQ', 'Marks', 'Package'])
df1
````

````python
# Method 2: From a dictionary (most common)
student_dict = {
    'IQ': [100, 90, 120],
    'Marks': [80, 70, 100],
    'Package': [10, 7, 14]
}
df2 = pd.DataFrame(student_dict)
df2
````

````python
# Method 3: From CSV (our main dataset for this notebook)
emp = pd.read_csv('employees.csv')
emp
````

## 5. The `axis` Parameter (Not in Series)

Series is 1D, so `s.sum()` has only one meaning. DataFrame is 2D, so you must specify **direction**:

| `axis` | Direction | Meaning |
|---|---|---|
| `axis=0` (default) | Top to bottom | | Operate on each **column** |
| `axis=1` | Left to right -> | Operate on each **row** |

> **Memory trick:** `axis=0` collapses rows (result has one value per column). `axis=1` collapses columns (result has one value per row).


````python
# axis=0 (default): sum of each column (top to bottom)
emp[['age', 'salary', 'experience_years']].sum()
````

````python
# axis=1: sum across columns for each row (left to right)
emp[['age', 'salary']].sum(axis=1).head()
````

## 6. Extracting Data

### Selecting Columns


````python
# Single column -> Series
emp['name']
````

````python
# Multiple columns -> DataFrame (double bracket = list inside bracket)
emp[['name', 'department', 'salary']]
````

### Selecting Rows: `iloc` vs `loc`

| | Series | DataFrame |
|---|---|---|
| Position | `s.iloc[0]` | `df.iloc[0]` (entire row as Series) |
| Label | `s.loc['Maths']` | `df.loc[0]` (by index label) |
| Slicing | `s.iloc[1:3]` | `df.iloc[0:5]` |
| Label slicing | `s.loc['A':'C']` (end included) | `df.loc[0:4]` (end included) |
| **2D access** | Not possible | `df.iloc[rows, cols]`, `df.loc[rows, cols]` |

> **Rule:** `iloc` - end is **excluded**. `loc` - end is **included**. Same rule as Series.


````python
# iloc - position-based (end excluded)
emp.iloc[0]    # first row as Series
````

````python
# iloc slicing
emp.iloc[0:5]    # first 5 rows (0,1,2,3,4)
````

````python
# iloc 2D: rows AND columns by position
emp.iloc[0:5, 0:4]    # first 5 rows, first 4 columns
````

````python
# loc - label-based (end included)
emp.loc[0:4]    # rows with index label 0 through 4 (inclusive)
````

````python
# loc 2D: rows by label, columns by name
emp.loc[0:4, ['name', 'department', 'salary']]
````

````python
# Fancy indexing - select specific rows
emp.iloc[[0, 5, 10, 15]]
````

## 7. Filtering (Boolean Indexing)

The concept is exactly the same as Series. The condition is applied on a **column (Series)**, and the filter returns matching **rows from the entire table**.

```python
# Series:   vk[vk >= 50]
# DataFrame: emp[emp['salary'] > 80000]   - condition on column, result is rows
```

**Multiple conditions:** Use `&` (AND), `|` (OR), `~` (NOT). Wrap each condition in **parentheses**.

> **Important:** Do NOT use Python's `and` / `or` / `not`. These don't work on Series.


````python
# Single condition: employees with salary > 80000
emp[emp['salary'] > 80000]
````

````python
# Multiple conditions: salary > 70000 AND department is IT
emp[(emp['salary'] > 70000) & (emp['department'] == 'IT')]
````

````python
# OR condition: department is HR or Finance
emp[(emp['department'] == 'HR') | (emp['department'] == 'Finance')]
````

````python
# NOT condition: employees NOT in IT
emp[~(emp['department'] == 'IT')]
````

````python
# String filtering with .str methods and apply
# Find employees who have 'Java' in their skills
mask = emp['skills'].str.contains('Java')
emp[mask]
````

````python
# Combined: Java skill AND salary > 60000
mask1 = emp['skills'].str.contains('Java')
mask2 = emp['salary'] > 60000
emp[mask1 & mask2]
````

## 8. Adding and Editing Columns

Adding a new column to a DataFrame is like adding a new label to a Series (`marks['SST'] = 90`), but here a new **column** is created.


````python
# Add a constant column (broadcasting)
emp['country'] = 'India'
emp.head()
````

## 9. Memory Optimization with `astype()`

`astype` is a Series method applied on DataFrame columns:

| Original dtype | Convert to | When to use |
|---|---|---|
| `int64` | `int32` / `int16` | When values are small (IDs, ages) |
| `object` (string) | `category` | When same strings repeat many times (departments, cities) |

> **Tip:** `category` type is ideal for columns where the same values appear many times (like department names, gender, grades).


````python
# Check current memory usage
print('Before optimization:')
emp.info(memory_usage='deep')
````

````python
# Convert 'department' and 'gender' to category type
emp_opt = emp.copy()
emp_opt['department'] = emp_opt['department'].astype('category')
emp_opt['gender'] = emp_opt['gender'].astype('category')
emp_opt['remote'] = emp_opt['remote'].astype('category')

print('\nAfter optimization:')
emp_opt.info(memory_usage='deep')
````

## 11. `sort_values()`

```python
# Series: s.sort_values(ascending=False)           - no 'by' needed
# DataFrame: df.sort_values('column', ascending=False)  - must specify column
```

**Key difference:** Series has only one column, so no `by` parameter needed. DataFrame **must** specify which column to sort by.

**Multi-column sorting** is DataFrame-only: sort by first column, then break ties with second column.


````python
# Sort by salary (descending)
emp.sort_values('salary', ascending=False).head(10)
````

````python
# Multi-column sort: by department (A-Z), then by salary (highest first)
emp.sort_values(['department', 'salary'], ascending=[True, False]).head(15)
````

````python
# NaN position control
emp.sort_values('salary', na_position='first').head()
````

## 12. `rank()`

The actual ranking happens on a Series (column). The result is stored as a new column in the DataFrame.

## 17. `dropna()` vs `fillna()`

### `dropna()` - Remove rows with missing values

```python
s.dropna()                          # Series: remove NaN entries
df.dropna()                         # DataFrame: how='any' (any NaN in row -> row removed)
df.dropna(how='all')                # Only if entire row is NaN
df.dropna(subset=['col1','col2'])   # Only check these columns for NaN
```

> `how` and `subset` are **DataFrame-only** parameters (Series has only one column).


````python
# dropna() - default drops row if ANY column has NaN
print('Before:', emp.shape)
emp_clean = emp.dropna()
print('After dropna():', emp_clean.shape)
````

````python
# dropna with subset - only check specific columns
emp_salary_clean = emp.dropna(subset=['salary', 'age'])
print('After dropna(subset=[salary, age]):', emp_salary_clean.shape)
````

### `fillna()` - Replace missing values

```python
s.fillna(0)                          s.fillna(s.mean())
df['age'].fillna(df['age'].mean())   # fill one column with its mean
df.ffill()                           df.bfill()
```

> Warning: `fillna(method='ffill')` is **deprecated** in newer Pandas. Use `ffill()` and `bfill()` directly.


````python
# fillna with a constant
emp['grade'].fillna('Unknown').value_counts()
````

````python
# fillna with mean (for numeric columns)
emp['age'].fillna(emp['age'].mean()).describe()
````

````python
# fillna with median (better for skewed data)
emp['salary'].fillna(emp['salary'].median()).describe()
````

````python
# ffill and bfill (Pandas 2.0+ safe)
emp['rating'].ffill().head(20)
````

````python
# Subset: check duplicates only by 'name' column
emp[emp.duplicated(subset=['name'], keep=False)].sort_values('name').head(10)
````

## 20. `apply()`

| Context | What it traverses |
|---|---|
| On Series | Each **element** |
| On DataFrame (`axis=0`) | Each **column** (default) |
| On DataFrame (`axis=1`) | Each **row** |

> **Key insight:** `axis=1` gives each row as a Series (with column names as index), so you can access multiple columns inside the function. This is **not possible in Series** since it has only one column.


````python
# apply on Series (each element)
emp['name'].apply(lambda x: x.split()[0]).head()
````

````python
# apply on DataFrame axis=0 (each column) - rarely used
emp[['age', 'salary', 'experience_years']].apply(lambda col: col.max() - col.min())
````

````python
# apply on DataFrame axis=1 (each row) - very useful!
# Create a summary string for each employee
def employee_summary(row):
    return f"{row['name']} ({row['department']}) - Rs {row['salary']}"

emp.apply(employee_summary, axis=1).head()
````

````python
# Practical example: compute experience-to-salary ratio
emp['efficiency'] = emp.apply(
    lambda row: row['salary'] / row['experience_years'] 
    if row['experience_years'] > 0 else 0, axis=1
)
emp[['name', 'salary', 'experience_years', 'efficiency']].head(10)
````

````python
# Cleanup
emp.drop(columns=['efficiency'], inplace=True)
````
